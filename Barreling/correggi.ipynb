### Image correction

```python
import numpy as np
import cv2
import discorpy.post.postprocessing as post
import discorpy.losa.loadersaver as losa
import matplotlib.pyplot as plt
```

Load the img to be corrected and its pedestal (ped) - needed for unvignetting.

```python
img = np.load('/private/camera/ORDINE/Vignetting/chess.npy')
ped = np.load('/private/camera/ORDINE/Vignetting/ped_chess.npy')
```

Load the vignetting map and the unbarrelling coefficients.

```python
muon_unvig = np.load('/private/muon/muon_unvig.npy')
muon_preproces = np.load('/private/muon/muon_preproces.npy')
```

```python
xcenter, ycenter, list_fact = losa.load_metadata_txt('/private/camera/ORDINE/Barrelling/output_discorpy/correction_coeffs.txt')
correction_map = np.load('/private/camera/ORDINE/Vignetting/output_dir/correction_map.npy')
```

```python
plt.imshow(muon_unvig, vmin=0, vmax=3)
plt.title('Original image')
plt.axis('off')
#plt.colorbar()
plt.show()
```

```python
unwarped = post.unwarp_image_backward(muon_unvig, xcenter, ycenter, list_fact)
plt.imshow(unwarped, vmin=0, vmax=3)
#plt.imshow(muon_unvig, vmin=0, vmax=3, alpha=0.3)
plt.title('Unbarrelled')
plt.axis('off')
plt.show()
```

```python
def acceptance(img,rowmin = 0,rowmax = 2304,colmin = 896,colmax = 4096-896):
    img[:rowmin,:]=0
    img[rowmax:,:]=0
    img[:,:colmin]=0
    img[:,colmax:]=0
    return img
```

```python
fig, ax = plt.subplots(1,2, figsize=(10,10), dpi=150)
#ax[0].imshow(acceptance(muon_preproces), vmin= 0, vmax= 3, cmap='viridis')
ax[0].imshow(acceptance(muon_unvig), vmin=0, vmax=3, cmap='viridis')
ax[1].imshow(acceptance(unwarped), vmin=0, vmax=3, cmap='viridis')
#ax[0].axis('off')
ax[0].axis('off')
ax[1].axis('off')
#ax[0].set_title('Original')
ax[0].set_title('Unvignetted')
ax[1].set_title('Unbarrelled')
plt.savefig('/private/muon/corrected.png')
```

```python
plt.figure(dpi=150)
plt.imshow(muon_unvig-unwarped, vmin=0, vmax=0.7)
plt.axis('off')
plt.savefig('/private/muon/diff.png')
```

```python
plt.figure(dpi=150)
plt.imshow(acceptance(muon_preproces), vmin=0, vmax=3)
plt.axis('off')
plt.title('Original')
plt.savefig('/private/muon/original.png')
```
